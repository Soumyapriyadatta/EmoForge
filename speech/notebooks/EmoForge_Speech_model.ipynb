```python
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, models, regularizers
from tensorflow.keras.callbacks import ModelCheckpoint, EarlyStopping, ReduceLROnPlateau
import matplotlib.pyplot as plt
import os
from sklearn.utils.class_weight import compute_class_weight
from sklearn.metrics import confusion_matrix, classification_report
import seaborn as sns
import librosa
from sklearn.model_selection import train_test_split

RAVDESS_DIR = '/content/drive/MyDrive/EmoForge/EmoForge/speech/data/Audio_Speech_Actors_01-24'
CREMA_DIR   = '/content/drive/MyDrive/EmoForge/EmoForge/speech/data/CREMA-D'
TESS_DIR    = '/content/drive/MyDrive/EmoForge/EmoForge/speech/data/TESS'
SAVE_DIR  = '/content/drive/MyDrive/EmoForge/EmoForge/speech/data/processed'
MODEL_DIR = '/content/drive/MyDrive/EmoForge/EmoForge/speech/models'
os.makedirs(SAVE_DIR, exist_ok=True)
os.makedirs(MODEL_DIR, exist_ok=True)

RAVDESS_MAP = {'01':4,'02':4,'03':3,'04':5,'05':0,'06':2,'07':1,'08':6}
CREMA_MAP   = {'ANG':0,'DIS':1,'FEA':2,'HAP':3,'NEU':4,'SAD':5}
TESS_MAP    = {'angry':0,'disgust':1,'fear':2,'happy':3,'neutral':4,'sad':5,'ps':6}

def extract_features(file_path, max_len=128, n_mfcc=40, n_mels=128):
    y, sr = librosa.load(file_path, sr=22050)
    mfcc   = librosa.feature.mfcc(y=y, sr=sr, n_mfcc=n_mfcc)
    mel    = librosa.power_to_db(librosa.feature.melspectrogram(y=y, sr=sr, n_mels=n_mels), ref=np.max)
    chroma = librosa.feature.chroma_stft(y=y, sr=sr)
    features = np.vstack([mfcc, mel, chroma])
    if features.shape[1] < max_len:
        features = np.pad(features, ((0,0),(0, max_len - features.shape[1])))
    else:
        features = features[:, :max_len]
    return features

X, y_list = [], []

print("Processing RAVDESS...")
for actor in sorted(os.listdir(RAVDESS_DIR)):
    actor_path = os.path.join(RAVDESS_DIR, actor)
    if not os.path.isdir(actor_path): continue
    for fname in os.listdir(actor_path):
        if not fname.endswith('.wav'): continue
        code = fname.split('-')[2]
        if code not in RAVDESS_MAP: continue
        try:
            X.append(extract_features(os.path.join(actor_path, fname)))
            y_list.append(RAVDESS_MAP[code])
        except Exception as e: print(f"Skipping {fname}: {e}")
print(f"After RAVDESS: {len(X)} samples")

print("Processing CREMA-D...")
for fname in os.listdir(CREMA_DIR):
    if not fname.endswith('.wav'): continue
    code = fname.split('_')[2]
    if code not in CREMA_MAP: continue
    try:
        X.append(extract_features(os.path.join(CREMA_DIR, fname)))
        y_list.append(CREMA_MAP[code])
    except Exception as e: print(f"Skipping {fname}: {e}")
print(f"After CREMA-D: {len(X)} samples")

print("Processing TESS...")
for folder in os.listdir(TESS_DIR):
    folder_path = os.path.join(TESS_DIR, folder)
    if not os.path.isdir(folder_path): continue
    code = folder.split('_')[-1].lower()
    if code not in TESS_MAP: continue
    for fname in os.listdir(folder_path):
        if not fname.endswith('.wav'): continue
        try:
            X.append(extract_features(os.path.join(folder_path, fname)))
            y_list.append(TESS_MAP[code])
        except Exception as e: print(f"Skipping {fname}: {e}")
print(f"After TESS: {len(X)} samples")

X = np.array(X)
y = np.array(y_list)
print(f"Full dataset: {X.shape}")

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)
np.save(f'{SAVE_DIR}/X_train.npy', X_train)
np.save(f'{SAVE_DIR}/y_train.npy', y_train)
np.save(f'{SAVE_DIR}/X_test.npy',  X_test)
np.save(f'{SAVE_DIR}/y_test.npy',  y_test)
print(f"Saved — X_train: {X_train.shape}, X_test: {X_test.shape}")

EMOTIONS = ['angry', 'disgust', 'fear', 'happy', 'neutral', 'sad', 'surprise']
NUM_CLASSES = 7

# Reshape
X_train = X_train.transpose(0, 2, 1)
X_test  = X_test.transpose(0, 2, 1)
print(f"Reshaped — X_train: {X_train.shape}")

# Augmentation (4x)
def augment_features(X, freq_mask=10, time_mask=10):
    X_aug = X.copy()
    for i in range(len(X_aug)):
        t  = np.random.randint(0, time_mask)
        t0 = np.random.randint(0, X_aug.shape[1] - t)
        X_aug[i, t0:t0+t, :] = 0
        f  = np.random.randint(0, freq_mask)
        f0 = np.random.randint(0, X_aug.shape[2] - f)
        X_aug[i, :, f0:f0+f] = 0
    return X_aug

X_train_aug1 = augment_features(X_train, freq_mask=8,  time_mask=8)
X_train_aug2 = augment_features(X_train, freq_mask=12, time_mask=15)
X_train_aug3 = augment_features(X_train, freq_mask=5,  time_mask=20)
X_train = np.concatenate([X_train, X_train_aug1, X_train_aug2, X_train_aug3], axis=0)
y_train = np.concatenate([y_train, y_train, y_train, y_train], axis=0)

# Train/Val split
X_tr, X_val, y_tr, y_val = train_test_split(
    X_train, y_train, test_size=0.15, stratify=y_train, random_state=42
)

# One-hot encode
y_tr_oh  = tf.keras.utils.to_categorical(y_tr,  num_classes=7)
y_val_oh = tf.keras.utils.to_categorical(y_val, num_classes=7)

# Class weights
class_weights = compute_class_weight('balanced', classes=np.unique(y_tr), y=y_tr)
class_weight_dict = dict(enumerate(class_weights))
print("Class weights:", class_weight_dict)

# Build CNN + BiLSTM model
def build_speech_model(input_shape, num_classes=7):
    inputs = layers.Input(shape=input_shape)

    x = layers.Conv1D(64, 3, padding='same', activation='relu',
        kernel_regularizer=regularizers.l2(1e-4))(inputs)
    x = layers.BatchNormalization()(x)
    x = layers.Conv1D(64, 3, padding='same', activation='relu',
        kernel_regularizer=regularizers.l2(1e-4))(x)
    x = layers.BatchNormalization()(x)
    x = layers.MaxPooling1D(2)(x)
    x = layers.Dropout(0.3)(x)

    x = layers.Conv1D(128, 3, padding='same', activation='relu',
        kernel_regularizer=regularizers.l2(1e-4))(x)
    x = layers.BatchNormalization()(x)
    x = layers.Conv1D(128, 3, padding='same', activation='relu',
        kernel_regularizer=regularizers.l2(1e-4))(x)
    x = layers.BatchNormalization()(x)
    x = layers.MaxPooling1D(2)(x)
    x = layers.Dropout(0.3)(x)

    x = layers.Bidirectional(layers.LSTM(128, return_sequences=True,
        kernel_regularizer=regularizers.l2(1e-4)))(x)
    x = layers.BatchNormalization()(x)
    x = layers.Dropout(0.3)(x)
    x = layers.Bidirectional(layers.LSTM(64, return_sequences=False,
        kernel_regularizer=regularizers.l2(1e-4)))(x)
    x = layers.BatchNormalization()(x)
    x = layers.Dropout(0.3)(x)

    x = layers.Dense(128, activation='relu',
        kernel_regularizer=regularizers.l2(1e-4))(x)
    x = layers.Dropout(0.4)(x)
    x = layers.Dense(64, activation='relu',
        kernel_regularizer=regularizers.l2(1e-4))(x)
    x = layers.Dropout(0.3)(x)
    outputs = layers.Dense(num_classes, activation='softmax')(x)

    return models.Model(inputs, outputs)

model = build_speech_model(input_shape=(X_tr.shape[1], X_tr.shape[2]))
model.summary()

# Compile
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss=tf.keras.losses.CategoricalCrossentropy(label_smoothing=0.1),
    metrics=['accuracy']
)

# Callbacks
checkpoint = ModelCheckpoint(
    f'{MODEL_DIR}/speech_model_best.h5',
    monitor='val_accuracy', save_best_only=True, verbose=1
)
early_stop = EarlyStopping(
    monitor='val_accuracy', patience=20,
    restore_best_weights=True, verbose=1
)
reduce_lr = ReduceLROnPlateau(
    monitor='val_loss', factor=0.5,
    patience=8, min_lr=1e-6, verbose=1
)

# Train
print("Starting training...")
history = model.fit(
    X_tr, y_tr_oh,
    validation_data=(X_val, y_val_oh),
    epochs=200,
    batch_size=32,
    class_weight=class_weight_dict,
    callbacks=[checkpoint, early_stop, reduce_lr]
)

# Evaluate
y_test_oh = tf.keras.utils.to_categorical(y_test, num_classes=7)
print("\nEvaluating on test set...")
test_loss, test_acc = model.evaluate(X_test, y_test_oh)
print(f"Test Accuracy: {test_acc*100:.2f}%")
print(f"Test Loss:     {test_loss:.4f}")

# Plot
plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
plt.plot(history.history['accuracy'], label='Train Accuracy')
plt.plot(history.history['val_accuracy'], label='Val Accuracy')
plt.title('Speech Model — Accuracy')
plt.xlabel('Epoch'); plt.ylabel('Accuracy'); plt.legend()

plt.subplot(1, 2, 2)
plt.plot(history.history['loss'], label='Train Loss')
plt.plot(history.history['val_loss'], label='Val Loss')
plt.title('Speech Model — Loss')
plt.xlabel('Epoch'); plt.ylabel('Loss'); plt.legend()

plt.tight_layout()
plt.savefig(f'{MODEL_DIR}/speech_model_training_history.png')
plt.show()
print("Training history plot saved!")

# Confusion Matrix
y_pred = np.argmax(model.predict(X_test), axis=1)
cm = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Oranges',
            xticklabels=EMOTIONS, yticklabels=EMOTIONS)
plt.title('Speech Model — Confusion Matrix')
plt.ylabel('True Label'); plt.xlabel('Predicted Label')
plt.tight_layout()
plt.savefig(f'{MODEL_DIR}/speech_model_confusion_matrix.png')
plt.show()
print("Confusion matrix saved!")

# Classification Report
report = classification_report(y_test, y_pred, target_names=EMOTIONS, output_dict=True)
print("\nClassification Report:")
print(classification_report(y_test, y_pred, target_names=EMOTIONS))

weighted_f1   = report['weighted avg']['f1-score']
best_emotion  = max(EMOTIONS, key=lambda e: report[e]['f1-score'])
worst_emotion = min(EMOTIONS, key=lambda e: report[e]['f1-score'])
best_epoch    = np.argmax(history.history['val_accuracy']) + 1

print(f"\nFINAL TEST ACCURACY: {test_acc*100:.2f}%")
print(f"WEIGHTED F1 SCORE: {weighted_f1:.4f}")
print(f"BEST PERFORMING EMOTION:  {best_emotion} — F1: {report[best_emotion]['f1-score']:.4f}")
print(f"WORST PERFORMING EMOTION: {worst_emotion} — F1: {report[worst_emotion]['f1-score']:.4f}")
print(f"NUMBER OF EPOCHS TRAINED: {len(history.history['loss'])}")
print(f"BEST EPOCH: {best_epoch}")
print(f"OPTIMIZER: Adam | LR: 0.001 | BATCH SIZE: 32")
print("\nSpeech model training complete! ✅")
```
